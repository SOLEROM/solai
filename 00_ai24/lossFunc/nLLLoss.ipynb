# NLLLoss - negative log-likelihood loss


Used for: Classification tasks (predicting categorical labels).

* where the output is a class label from a set of possible categories.

!!! NLLLoss expects the model's output to be log probabilities obtained after applying a log softmax activation !!!



- **Formula:**
  $$
  NLLLoss = -\sum_{i=1}^{n} \log(p_{y_i})
  $$
  where $ p_{y_i} $ is the predicted probability (already in log form) of the true class $ y_i $, and $ n $ is the number of samples.


* NLL measures how well the predicted probabilities align with the true class labels. The log function penalizes predictions that assign low probabilities to the correct class.
* The closer the predicted probability for the true class is to 1, the smaller the loss. Conversely, if the probability is near zero, the loss is very large.
* Objective: Minimize the NLL to improve the confidence in correct predictions. This loss function is commonly used in classification scenarios with softmax output layers.
